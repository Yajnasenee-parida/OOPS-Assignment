1. **Class**:

   A class in Python is a user-defined template that bundles data (attributes) and functions (methods) together, allowing you to model real-world entities in code. Classes are created using the class keyword.

   **Object**:

   An object is a specific instance of a class. Each object has its own set of data and can invoke the class methods.


```python
# 1
class Dog:
    species = "Canis familiaris"  # Class attribute
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        return f"{self.name} says woof!"


dog1 = Dog("Buddy", 3)      # dog1 and dog2 are object
dog2 = Dog("Max", 5)
print(dog1.bark())
print(dog2.bark())
```

2.   The four pillars of OOP are abstraction, encapsulation, inheritance, and polymorphism, which together enable modular, reusable, and maintainable software design.

3.   The __init__() function is used in Python classes to initialize the state of an object when it is created. It is automatically called when a new object of a class is instantiated, allowing the programmer to set up the object's attributes and ensure that each object starts with a consistent state.


```python
# 3
class Person:
   def __init__(self, name, age):
       self.name = name
       self.age = age

   def return_person_details(self):
       return self.name , self.age

obj = Person("Yajna" , 25)
obj.return_person_details()
```

4.   In Python’s object-oriented programming, self is a reference to the current instance of a class. It is explicitly included as the first parameter in instance methods so that the method can access or modify the object’s attributes and call other methods on the same object.



5.   Inheritance in Python is an Object-Oriented Programming (OOP) concept where a child (derived) class can acquire the attributes and methods of a parent (base) class.
It promotes code reusability, extensibility, and maintainability.



```python
# 5

# Single Inheritance Example
class Parent:
    def greet(self):
        print("Hello from Parent class!")

class Child(Parent):
    def welcome(self):
        print("Welcome from Child class!")

obj = Child()
obj.greet()
obj.welcome()
```

```python
# 5

# Multiple Inheritance Example
class Father:
    def skill(self):
        print("Father: Knows driving")

class Mother:
    def hobby(self):
        print("Mother: Loves cooking")

class Child(Father, Mother):
    def talent(self):
        print("Child: Plays guitar")

obj = Child()
obj.skill()
obj.hobby()
obj.talent()
```
